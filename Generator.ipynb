```python
# Generator Function
def gen(x):
    x = x+1
    yield x
    x = x+2
    yield x
    x = x+3
    yield x
res = gen(5)
print(res)
print(next(res))
print(next(res))
print(next(res))
```

```python
# Generator Definition
def gen(x):
    x = x+1
    yield x
    x = x+2
    yield x
    x = x+3
    yield x
res = gen(5)
for i in res:
    print(i)
```
