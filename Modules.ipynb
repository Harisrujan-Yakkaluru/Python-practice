```python
import random
print(random.random()) # This funtion returns the value in between o to 1
print(random.randint(1, 10)) # This function generates random number between 
print(random.choice(['A', 1, 2, 3, 4, 5, 6, 7, 8, 9, 'J', 'Q', 'K'])) #It returns random item from the sequence
print(random.choices(['A', 1, 2, 3, 4, 5, 6, 7, 8, 9, 'J', 'Q', 'K'], k=3))
arr = (['A', 1, 2, 3, 4, 5, 6, 7, 8, 9, 'J', 'Q', 'K'])
print(arr)
```
