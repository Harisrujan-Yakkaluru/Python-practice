```python
# Updating Instance attribute data from outside class
class Mobile:
    company = "Iphone"
    def __init__(self, model, ram, rom, price):
        self.model = model
        self.ram = ram
        self.rom = rom
        self.price = price
    def about_phone(self):
        return f"The {self.model} have {self.ram} GB RAM and {self.rom} GB ROM"
    def display():
        return "This is Samsung Company Display"
    # Class Method
    @classmethod
    def update_company(cls, new_name):
        cls.company = new_name
        return "Company name updated"
    # Static Method
    @staticmethod
    def charger_provide(cable=False):
        return "Adapter + Cable" if cable else "Only Adapter"
# Object creation
# Object creation
iphone16 = Mobile("iphone 16", 16, 256, 125000)
iphone17 = Mobile("iphone 17", 8, 256, 120000)
print(iphone16.about_phone())
print(iphone16.update_company("Iphone Pro"))
print(iphone16.charger_provide())
print(iphone17.company)
```

```python
class Account:
    Bank_name = "Harisrujan"
    def __init__(self, user_name, balance, password):
        self.user_name = user_name # Public
        self._balance = balance # Protected
        self.__password = password # Private
    def get_balance(self): # Getter Method
        return self._balance
    def withdraw(self, withdraw_amount):
        if self._balance >= withdraw_amount:
            self._balance -= amount
            return f"{withdraw_amount} withdrawl successful"
        return "Insufficient Balance"
    def deposit(self, deposit_amount): # Setter Method
        self._balance += deposit_amount
        return f"{deposit_amount} Deposited Successful and Current Balance is {self._balance}"
    def __total_users(self): # Private Method
        return "Total Users: 100"
user = Account("Hari", 8000, "Hari123")
```

```python
# Update Password
class Account:
    Bank_name = "Harisrujan"
    def __init__(self, user_name, balance, password):
        self.user_name = user_name # Public
        self._balance = balance # Protected
        self.__password = password # Private
    def get_balance(self): # Getter Method
        return self._balance
    def update_password(self, new_password):
        self.__password = new_password
        return "Password Updated"
    def get_password(self):
        return self.__password
    def withdraw(self, withdraw_amount):
        if self._balance >= withdraw_amount:
            self._balance -= amount
            return f"{withdraw_amount} withdrawl successful"
        return "Insufficient Balance"
    def deposit(self, deposit_amount): # Setter Method
        self._balance += deposit_amount
        return f"{deposit_amount} Deposited Successful and Current Balance is {self._balance}"
    def __total_users(self): # Private Method
        return "Total Users: 100"
user = Account("Hari", 8000, "Hari123")
print(user.user_name)
print(user.get_balance())
print(user.deposit(5000))
print(user.update_password("Srujan123"))
print(user.get_password())

# Access the protected member syntax: obj._attributename
print(user._balance)
# Access private member by using name mangling
# Syntax: obj._classname__attributename
print(user._Account__total_users())
```
