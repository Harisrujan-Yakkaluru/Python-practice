```python
# Method Overriding
class Person:
    def __init__(self, name:str, age:int):
        self.name = name
        self.age = age
    def details(self):
        return f"My name is {self.name} and I am a {self.age} years old"
class Student(Person):
    def __init__(self, name:str, age:int, roll_no:int):
        super().__init__(name, age)
        self.roll_no = roll_no
    def details(self):
        return f"I am a student"
p = Person("ravi", 15)
s = Student("srinu", 14, 1001)
print(p.details())
print(s.details())
```

```python
# Method Overloading
class Math:
    def add(self, a, b):
        return a + b
    def add(self, a, b, c):
        return a + b + c
    def add(self, a, b, c = 0, d = 0):
        return a + b + c + d
x = Math()
print(x.add(5, 5))
print(x.add(1, 2, 3))
print(x.add(7, 9, 3, 5))
```

```python
# Operator Overloading
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
p1 = Point(5, 3)
print(p1)
```

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def __str__(self): # String Magic Method
        return f"This point class and value is : ({self.x}, {self.y})"
p1 = Point(5, 3)
print(p1)
```

```python
class Point:
    def __init__(self, x):
        self.x = x
    def __str__(self): # String Magic Method
        return f"This point class and value is : {self.x}"
p1 = Point(5)
p2 = Point(3)
print(p1.x + p2.x)
```

```python
class Point:
    def __init__(self, x):
        self.x = x
    def __str__(self): # String Magic Method
        return f"This point class and value is : {self.x}"
    def __add__(self, other):
        return self.x + other.x
p1 = Point(5)
p2 = Point(3)
print(p1.x + p2.x)
print(p1+p2)
```

```python
class Point:
    def __init__(self, x):
        self.x = x
    def __str__(self): # String Magic Method
        return f"This point class and value is : {self.x}"
    def __add__(self, other):
        return self.x + other.x
    def __sub__(self, other):
        return self.x - other.x
p1 = Point(5)
p2 = Point(3)
print(p1.x + p2.x)
print(p1+p2)
print(p1-p2)
```
