```python
# 1. Single Inheritance
# With implementation in child class
# When child class don't have constructor, in this case parent class constructor behave like child class constructor
# Parent Class
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def details(self):
        return f"My name is {self.name} and I am a {self.age} years old"
# Child Class
class Student(Person):
    pass
s = Student("ravi", 20)
print(s.name)
print(s.details())
```

```python
# Parent Class
class Person:
    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender
    def details(self):
        return f"My name is {self.name} and I am a {self.age} years old and {self.gender}"
# Child Class
class Student(Person):
    def __init__(self, stdid, name, age):
        self.stdid = stdid
        self.name = name
        self.age = age
p = Person("Raja", 21, "Male")
s = Student(101, "ravi", 20)
print(s.name)
s.gender = "Male"
print(p.details())
print(s.details())
```

```python
# Inherit the parent class constructor in child class 
class Person:
    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender
    def details(self):
        return f"My name is {self.name} and I am a {self.age} years old and {self.gender}"
# Child Class
class Student(Person):
    def __init__(self, stdid, name, age, gender):
        self.stdid = stdid
        super().__init__(name, age, gender)
    def percentage(self, marks:list):
        return sum(marks)/len(marks)

s = Student(101, "ravi", 20, "male")
print(s.name)
print(s.details())
print(s.percentage([90, 89, 88]))
```

```python
class employee:
    def __init__(self, emp_id:int, name:str, salary:int):
        self.emp_id = emp_id
        self.name = name
        self.salary = salary
    def calculate_salary(self, bonus=0):
        return self.salary + bonus
class team_lead(employee):
    def __init__(self, emp_id:int, name:str, salary:int, project_name:str):
        self.project_name = project_name
        super().__init__(emp_id, name, salary)
    def project_status(self, status):
        return f"{self.project_name} Project Status: {status}% completed"
class manager(team_lead):
    def __init__(self, emp_id:int, name:str, salary:int, project_name:str, no_projects:int):
        self.no_projects = no_projects
        super().__init__(emp_id, name, salary, project_name)
a = employee(1, "Hari", 1000000)
b = team_lead(101, "Timothee", 1515515, "Dune")
c = manager(1111, "SpiderMan", 508080000, "BND", 4)
print(a.name, b.name, c.name)
print(a.calculate_salary())
print(b.calculate_salary(5000))
print(b.project_status(90))
print(c.calculate_salary(100000))
print(c.project_status(99999))
```

```python
# Multiple Inheritance
class Father:
    def __init__(self, father_name:str):
        self.father_name = father_name
    def working(self):
        return "Father is a Gladiator"
class Mother:
    def __init__(self, mother_name:str):
        self.mother_name = mother_name
    def working(self):
        return "Mother is a Radiator"
class Child:
    def __init__(self, father_name, mother_name, name:str):
        self.name = name
        Father.__init__(self, father_name)
        Mother.__init__(self, mother_name)
    def working(self):
        f = Father(self.father_name)
        m = Mother(self.mother_name)
        print(f.working())
        print(m.working())
c = Child("Maximus", "Decimus", "Meridius")
print(c.father_name)
print(c.mother_name)
print(Child.__mro__) # Method Resolution Order
print(c.working())
```

```python
# Multiple Inheritance
class Father:
    def __init__(self, father_name:str):
        self.father_name = father_name
    def working1(self):
        return "Father is a Gladiator"
class Mother:
    def __init__(self, mother_name:str):
        self.mother_name = mother_name
    def working2(self):
        return "Mother is a Radiator"
class Child:
    def __init__(self, father_name, mother_name, name:str):
        self.name = name
        Father.__init__(self, father_name)
        Mother.__init__(self, mother_name)
    def working(self):
        print(super().working1())
        print(super().working2())
c = Child("Maximus", "Decimus", "Meridius")
print(c.father_name)
print(c.mother_name)
print(Child.__mro__) # Method Resolution Order
print(c.working())
```
